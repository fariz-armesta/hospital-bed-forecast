```python
def population_data_cleaning(filename, year):
    data = "C:\\Coding\\hospital-bed-forecast\\data\\raw\\" + filename
    df = pd.read_csv(data)
    df = df.dropna()
    df = df.replace("...", 0)
    df = df.replace("–", 0)
    df["Jumlah Penduduk (Ribu)"] = df["Jumlah Penduduk (Ribu)"].astype(float)
    df["Jumlah Penduduk (Ribu)"] = df["Jumlah Penduduk (Ribu)"] * 1000
    df["Jumlah Penduduk (Ribu)"] = df["Jumlah Penduduk (Ribu)"].astype(int)
    df = df.rename(columns={
        "Jumlah Penduduk (Ribu)": ("Jumlah Penduduk " + year),
        "Laju Pertumbuhan Penduduk per Tahun": ("LPP " + year)
    })
    df = df.drop(columns=["Persentase Penduduk", "Kepadatan Penduduk per km persegi (Km2)", "Rasio Jenis Kelamin Penduduk"])
    df[("LPP " + year)] = df[("LPP " + year)].astype(str).str.replace(r'[^\d.-]', '', regex=True)
    df[("LPP " + year)] = df[("LPP " + year)].astype(float)

    return df
```

```python
data = "C:\\Coding\\hospital-bed-forecast\\data\\raw\\Laju Pertumbuhan Penduduk menurut Provinsi 2011 -2014.csv"
df_2011_2014 = pd.read_csv(data)
df_2011_2014.tail()
```

```python
for col in ["2011", "2012", "2013", "2014"]:
    df_2011_2014[col] = df_2011_2014[col].astype(str).str.replace(",", ".", regex=False).astype(float)

df_2011_2014["2010"] = df_2011_2014["2010"].astype(str).str.replace(" ", "", regex=False).astype(int)
df_2011_2014.head()
```

```python
for index, row in df_2011_2014.iterrows():
    pop_2011 = int(np.floor(row["2010"] + ((row["2010"] * row["2011"]) / 100)))
    pop_2012 = int(np.floor(pop_2011 + ((pop_2011 * row["2012"]) / 100)))
    pop_2013 = int(np.floor(pop_2012 + ((pop_2012 * row["2013"]) / 100)))
    pop_2014 = int(np.floor(pop_2013 + ((pop_2013 * row["2014"]) / 100)))
    
    df_2011_2014.at[index, "2011"] = pop_2011
    df_2011_2014.at[index, "2012"] = pop_2012
    df_2011_2014.at[index, "2013"] = pop_2013
    df_2011_2014.at[index, "2014"] = pop_2014

df_2011_2014 = df_2011_2014.drop(columns=["2010", "2011", "2012"])
df_2011_2014["2013"] = df_2011_2014["2013"].astype(int)
df_2011_2014["2014"] = df_2011_2014["2014"].astype(int)
df_2011_2014.head()
```

```python
data = "C:\\Coding\\hospital-bed-forecast\\data\\raw\\Population By Province in Indonesia (Thousand), 2015 -  2019.csv"
df_2015_2019 = pd.read_csv(data)
df_2015_2019.head()
```

```python
df_2015_2019 = df_2015_2019.rename(columns={'Provinsi/Province': 'Provinsi'})
df_2015_2019 = df_2015_2019.rename(columns={'2015': 'Jumlah Penduduk 2015'})
df_2015_2019["Jumlah Penduduk 2015"] = df_2015_2019["Jumlah Penduduk 2015"].astype(str).str.replace(',', '.', regex=False)
df_2015_2019["Jumlah Penduduk 2015"] = df_2015_2019["Jumlah Penduduk 2015"].astype(str).str.replace(" ", "", regex=False).astype(float)
df_2015_2019["Jumlah Penduduk 2015"] = df_2015_2019["Jumlah Penduduk 2015"] * 1000
df_2015_2019 = df_2015_2019.drop(columns=["2016", "2017", "2018", "2019"])
df_2015_2019.head()
```
